# AdventureWorksDW2022 ETL Pipeline Demonstration

This notebook demonstrates the complete data engineering pipeline that mirrors a cloud-scale Microsoft Fabric architecture on your local machine.

## 1. Setup and Imports

```python
# Import required libraries
import sys
import os
sys.path.append(os.path.join(os.getcwd(), 'src'))

from src.extract import DataExtractor
from src.transform import DataTransformer
from src.load import DataLoader
from utils.db_utils import DatabaseConnector
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

## 2. Extract Data

```python
# Initialize database connector and extractor
db = DatabaseConnector()
extractor = DataExtractor(db)

# Extract key tables
key_tables = [
    'DimCustomer',
    'DimProduct', 
    'DimDate',
    'DimGeography',
    'FactInternetSales'
]

print(f"Extracting {len(key_tables)} tables from AdventureWorksDW2022...")
raw_data = extractor.extract_all_tables(key_tables)

# Display extraction results
for table_name, df in raw_data.items():
    print(f"{table_name}: {len(df)} rows, {len(df.columns)} columns")
```

## 3. Transform Data

```python
# Initialize transformer
transformer = DataTransformer()

# Transform data
print("Transforming data...")
transformed_data = transformer.transform_all(raw_data)

# Display transformation results
for table_name, df in transformed_data.items():
    print(f"{table_name}: {len(df)} rows, {len(df.columns)} columns")
```

## 4. Load to Parquet Files

```python
# Initialize loader
loader = DataLoader()

# Save to Parquet files
print("Saving to Parquet files...")
saved_files = loader.save_to_parquet(transformed_data, version="v1")
print(f"Saved {len(saved_files)} files")
```

## 5. Validate Results

```python
# Display file information
print("Parquet File Details:")
for file_path in saved_files:
    if os.path.exists(file_path):
        info = loader.get_file_info(file_path)
        print(f"  {os.path.basename(file_path)}: {info.get('num_rows', 'N/A')} rows, "
              f"{info.get('file_size', 'N/A')} bytes")
```
